# Step 1: Dataset

For this project, I chose the NYC 311 Service Requests from 2020 to Present dataset. This dataset is published by NYC Open Data and is also available through public open-government data sources. It contains public service requests submitted by residents, including complaints about noise, illegal parking, sanitation, housing conditions, street issues, damaged trees, heat/hot water problems, and other city service concerns. The dataset meets the assignment requirement because it has far more than 500 rows and can be opened in Python as a CSV file using pandas. The data is public open-government data, so it can be used for educational and analysis purposes. An AI use case for this dataset is classification, where a model could learn to predict the urgency level of a service request, the responsible city agency, or the type of complaint based on the complaint description and related fields. One quality problem I noticed is that the dataset has missing values in fields such as closed date, descriptor, incident ZIP, city, latitude, and longitude. It also contains inconsistent formats in date fields and some incomplete location information, which makes it useful for practicing data cleaning and curation.

# Step 2a: Load and Count

```python
from IPython.display import display, HTML

display(HTML("""
<h1 style="background-color:#dbeafe; color:#1e3a8a; padding:15px; border-radius:10px; text-align:center;">
Step 2a: Load and Count
</h1>
"""))
```

```python
import pandas as pd
import numpy as np
from urllib.parse import urlencode

base_url = "https://data.cityofnewyork.us/resource/erm2-nwe9.csv"

params = {
    "$limit": 1000000,
    "$select": "unique_key,created_date,closed_date,agency,complaint_type,descriptor,incident_zip,city,status,borough,latitude,longitude"
}

url = base_url + "?" + urlencode(params)

df = pd.read_csv(url)

print("Rows before cleaning:", df.shape[0])
print("Columns before cleaning:", df.shape[1])
df.head()
```

# Step 2b: Profile the Data

```python
from IPython.display import display, HTML

display(HTML("""
<h1 style="background-color:#dbeafe; color:#1e3a8a; padding:15px; border-radius:10px; text-align:center;">
Step 2: Profile the Data
</h1>
"""))
```

```python
df.info()
df.describe(include="all")
print("Missing values in each column:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nUnique values in each column:")
print(df.nunique())
```

# Step 2c: Fixing Two Problems

```python
from IPython.display import display, HTML

display(HTML("""
<h1 style="background-color:#dbeafe; color:#1e3a8a; padding:15px; border-radius:10px; text-align:center;">
Step 2a:Fixing Two Problems
</h1>
"""))
```

```python
df_clean = df.copy()

# Fix 1: Fill missing values in important category/text columns
df_clean["descriptor"] = df_clean["descriptor"].fillna("Unknown")
df_clean["city"] = df_clean["city"].fillna("Unknown")
df_clean["incident_zip"] = df_clean["incident_zip"].fillna("Unknown")

# Fix 2: Standardize date columns
df_clean["created_date"] = pd.to_datetime(df_clean["created_date"], errors="coerce")
df_clean["closed_date"] = pd.to_datetime(df_clean["closed_date"], errors="coerce")

# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

print("Cleaning is complete.")
print("Rows after cleaning:", df_clean.shape[0])
print("Columns after cleaning:", df_clean.shape[1])
df_clean.head()
```

# Step 2 written explanation

I cleaned two main quality problems in the NYC 311 dataset. First, I filled missing values in important text and category columns such as descriptor, city, and incident ZIP with “Unknown” so the missing information is clearly marked instead of left blank. Second, I standardized the created date and closed date columns by converting them into datetime format. This makes the dates easier to analyze and helps avoid problems caused by inconsistent date formatting. I also removed exact duplicate rows to improve uniqueness. After cleaning, I recorded the row and column count again to show how the dataset changed.

# Step 3a — Annotation Schema

# Label name: Service Urgency

# Label options and rules:

#High urgency:
The complaint may affect safety, health, housing, sanitation, emergency access, or public danger. Examples include heat/hot water problems, electrical problems, animal abuse, fallen tree limbs, traffic signal issues, elevator problems, and plumbing problems.

#Medium urgency:
The complaint is important and may affect access, visibility, transportation, property use, or neighborhood safety, but it does not clearly show an immediate emergency. Examples include blocked driveways, street light problems, failed street repairs, food establishment permit issues, and overgrown branches.

# Low urgency:
The complaint is a routine service issue or quality-of-life concern with no immediate danger shown. Examples include routine noise complaints, posted parking violations, commercial overnight parking, and sanitation vehicle noise.

Needs review: The complaint does not include enough information to determine the urgency, or the descriptor is unclear or unknown.

Edge case rule: If I cannot decide the urgency level because the complaint description is too general or missing important details.#

# Key Take Away Report


The key takeaway from this project is that data curation is an important step before using any dataset for artificial intelligence or machine learning. A dataset may look complete at first, but after loading and profiling it, many issues can appear, such as missing values, inconsistent formats, unclear labels, duplicate records, or incomplete location information. This project helped me understand that AI models depend on clean, organized, and well-documented data.

For my project, I used the **NYC 311 Service Requests from 2020 to Present** dataset. This dataset is useful because it contains real public service complaints from residents, including issues related to noise, illegal parking, sanitation, housing, trees, streets, and other city services. One possible AI use case is to classify service requests by urgency level, such as **Low urgency, Medium urgency, High urgency, or Needs review**. This type of classification could help city agencies understand which complaints may need faster attention.

During the curation process, I encountered several issues. One issue was that some records had missing information in important columns such as **descriptor, city, incident ZIP, closed date, latitude, and longitude**. Another issue was that the date columns were not always ready for analysis, so they had to be converted into a standard datetime format. I also noticed that some complaint descriptions were too general or unclear, which made it harder to decide the correct urgency label. For example, some records did not provide enough detail to know whether the complaint should be labeled as low, medium, or high urgency. Because of this, I used the **Needs review** label for unclear cases.

I also learned that annotation requires clear rules. When labeling 50 rows, I had to create a label schema so that each urgency level had a clear meaning. This helped make the labeling process more consistent. However, some records were still difficult to label because real-world data can be messy and incomplete. This showed me that human judgment is still important in data curation.

Overall, this project showed me that data curation is not only a technical process, but also a responsible practice. Cleaning, labeling, scoring, and documenting the dataset made it easier to understand and use. The NYC 311 dataset is helpful for learning and analysis, but it should not be used as the only source for emergency decisions or unfair judgments about neighborhoods because 311 complaints may not represent every issue equally.
